```python
import pandas as pd
import numpy as np
from joblib import dump
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.svm import SVR
import lightgbm as lgb
import xgboost as xgb
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.font_manager import FontProperties
from matplotlib_inline.backend_inline import set_matplotlib_formats
from matplotlib.colors import LinearSegmentedColormap
from sklearn.inspection import PartialDependenceDisplay
import shap
import warnings
import os

set_matplotlib_formats('svg')
warnings.filterwarnings('ignore')


# -----------------------------------------------------------------------------
# FONT SETTINGS
# -----------------------------------------------------------------------------
FONT_FAMILY             = 'Arial'
TITLE_FONT_SIZE         = 14
TITLE_FONT_WEIGHT       = 'bold'
LABEL_FONT_SIZE         = 12
TICK_FONT_SIZE          = 10
VALUE_FONT_SIZE         = 10
LEGEND_FONT_SIZE        = 10

# -----------------------------------------------------------------------------
# GLOBAL MATPLOTLIB / SEABORN DEFAULTS
# -----------------------------------------------------------------------------
GLOBAL_FONT_SIZE        = 10
FIGURE_DPI              = 120
AXES_LINE_WIDTH         = 0.8
GRID_COLOR              = '#CCCCCC'
GRID_ALPHA              = 0.5
SNS_STYLE               = 'whitegrid'
SNS_PALETTE             = 'muted'
SNS_FONT_SCALE          = 1.0

# -----------------------------------------------------------------------------
# FIGURE SIZES  (width, height) in inches
# -----------------------------------------------------------------------------
TRUE_PRED_SCATTER_FIGSIZE       = (3.35, 2.91)   # 8.5 × 7.4 cm — separate Train & Test
RESIDUAL_ANALYSIS_TRAIN_FIGSIZE = (3.35, 2.91)
RESIDUAL_ANALYSIS_TEST_FIGSIZE  = (3.35, 2.91)
FEATURE_IMPORTANCE_FIGSIZE      = (3.35, 2.91)
SHAP_SUMMARY_PLOT_FIGSIZE       = (6.00, 6.50)
SHAP_BAR_PLOT_FIGSIZE           = (6.00, 6.50)
SHAP_WATERFALL_PLOT_FIGSIZE     = (3.35, 2.91)
SHAP_DEPENDENCE_GRID_FIGSIZE    = (15, 8)
PDP_GRID_FIGSIZE                = (14, 10)

# -----------------------------------------------------------------------------
# PDP SETTINGS
# -----------------------------------------------------------------------------
PDP_SELECTED_FEATURES   = ['Hf', 'U', 'Th', 'Nb']  # Zr excluded from dataset
PDP_LINE_COLOR          = '#2E86AB'
PDP_LINE_WIDTH          = 2.5

# -----------------------------------------------------------------------------
# SCATTER PLOT SETTINGS
# -----------------------------------------------------------------------------
SCATTER_TRAIN_COLOR     = '#b107fa'
SCATTER_TEST_COLOR      = '#CBB300'
SCATTER_SIZE            = 36
SCATTER_ALPHA           = 0.9
SCATTER_EDGE_WIDTH      = 0.2

TRUE_PRED_SCATTER_POINT_SIZE   = 22
TRUE_PRED_SCATTER_TITLE_SIZE   = 10
TRUE_PRED_SCATTER_LABEL_SIZE   = 8
TRUE_PRED_SCATTER_LABEL_WEIGHT = 'bold'
TRUE_PRED_SCATTER_TICK_SIZE    = 7
TRUE_PRED_SCATTER_VALUE_SIZE   = 7
TRUE_PRED_SCATTER_LEGEND_SIZE  = 7
TRUE_PRED_SCATTER_LINE_WIDTH   = 1.0
TRUE_PRED_SCATTER_TICK_LENGTH  = 2.5
TRUE_PRED_SCATTER_TICK_WIDTH   = 0.6

# -----------------------------------------------------------------------------
# RESIDUAL PLOT SETTINGS
# -----------------------------------------------------------------------------
RESIDUAL_HIST_BINS       = 20
RESIDUAL_HIST_COLOR      = '#00E936'
RESIDUAL_HIST_ALPHA      = 0.7
RESIDUAL_HIST_EDGE_COLOR = 'black'
RESIDUAL_HIST_EDGE_WIDTH = 0.4
ZERO_LINE_COLOR          = 'black'
ZERO_LINE_STYLE          = '--'
ZERO_LINE_WIDTH          = 1.1

RESIDUAL_POINT_SIZE      = 22
RESIDUAL_TITLE_SIZE      = 10
RESIDUAL_LABEL_SIZE      = 8
RESIDUAL_LABEL_WEIGHT    = 'bold'
RESIDUAL_TICK_SIZE       = 7
RESIDUAL_LEGEND_SIZE     = 7
RESIDUAL_TICK_LENGTH     = 2.5
RESIDUAL_TICK_WIDTH      = 0.6

# -----------------------------------------------------------------------------
# FEATURE IMPORTANCE SETTINGS
# -----------------------------------------------------------------------------
FEATURE_IMPORTANCE_BAR_HEIGHT     = 0.95
FEATURE_IMPORTANCE_BAR_ALPHA      = 0.95
FEATURE_IMPORTANCE_BAR_EDGE_COLOR = 'white'
FEATURE_IMPORTANCE_BAR_EDGE_WIDTH = 1
FEATURE_IMPORTANCE_COLORMAP       = 'viridis'
FEATURE_IMPORTANCE_CMAP_LOW       = 0.9
FEATURE_IMPORTANCE_CMAP_HIGH      = 0.3
FEATURE_IMPORTANCE_TOP_N          = 10
FEATURE_IMPORTANCE_XLIM_SCALE     = 1.2
FEATURE_IMPORTANCE_VALUE_OFFSET   = 1
FEATURE_IMPORTANCE_TITLE_SIZE     = 10
FEATURE_IMPORTANCE_TITLE_WEIGHT   = 'bold'
FEATURE_IMPORTANCE_LABEL_SIZE     = 8
FEATURE_IMPORTANCE_LABEL_WEIGHT   = 'bold'
FEATURE_IMPORTANCE_TICK_SIZE      = 7
FEATURE_IMPORTANCE_VALUE_SIZE     = 7
FEATURE_IMPORTANCE_TICK_LENGTH    = 2.5
FEATURE_IMPORTANCE_TICK_WIDTH     = 0.6

# -----------------------------------------------------------------------------
# SHAP SETTINGS
# -----------------------------------------------------------------------------
SHAP_COLORMAP             = 'viridis'
SHAP_SCATTER_SIZE         = 25
SHAP_SCATTER_ALPHA        = 0.7
SHAP_SCATTER_EDGE_COLOR   = 'white'
SHAP_SCATTER_EDGE_WIDTH   = 0.3
SHAP_WATERFALL_SAMPLE_IDX = 90
SHAP_TOP_N_FEATURES       = 5
SHAP_SUMMARY_TITLE_SIZE   = 20
SHAP_SUMMARY_TITLE_WEIGHT = 'bold'
SHAP_SUMMARY_LABEL_SIZE   = 16
SHAP_SUMMARY_LABEL_WEIGHT = 'bold'
SHAP_SUMMARY_TICK_SIZE    = 14

# -----------------------------------------------------------------------------
# INTERACTION BAR SETTINGS
# -----------------------------------------------------------------------------
INTERACTION_BAR_HEIGHT     = 0.7
INTERACTION_BAR_ALPHA      = 0.8
INTERACTION_BAR_EDGE_COLOR = 'black'
INTERACTION_BAR_EDGE_WIDTH = 0.3
INTERACTION_COLORMAP       = 'viridis'
INTERACTION_CMAP_LOW       = 0.3
INTERACTION_CMAP_HIGH      = 0.9

# -----------------------------------------------------------------------------
# AXES / TICK / BORDER SETTINGS
# -----------------------------------------------------------------------------
TICK_DIRECTION    = 'out'
TICK_LENGTH       = 3.5
TICK_WIDTH        = 0.8
TICK_COLOR        = 'black'
X_NBINS           = 6
Y_NBINS           = 6
BORDER_LINE_WIDTH = 0.5
BORDER_COLOR      = 'black'

# =============================================================================
# END OF PLOT SETTINGS
# =============================================================================

plt.rc('font', family=FONT_FAMILY)
plt.rcParams['font.size']      = GLOBAL_FONT_SIZE
plt.rcParams['figure.dpi']     = FIGURE_DPI
plt.rcParams['axes.linewidth'] = AXES_LINE_WIDTH
plt.rcParams['grid.color']     = GRID_COLOR
plt.rcParams['grid.alpha']     = GRID_ALPHA
sns.set(style=SNS_STYLE, palette=SNS_PALETTE, font_scale=SNS_FONT_SCALE)

title_font  = FontProperties(family=FONT_FAMILY, size=TITLE_FONT_SIZE,  weight=TITLE_FONT_WEIGHT)
label_font  = FontProperties(family=FONT_FAMILY, size=LABEL_FONT_SIZE)
tick_font   = FontProperties(family=FONT_FAMILY, size=TICK_FONT_SIZE)
value_font  = FontProperties(family=FONT_FAMILY, size=VALUE_FONT_SIZE)
legend_font = FontProperties(family=FONT_FAMILY, size=LEGEND_FONT_SIZE)

# =============================================================================
# DATASET / TRAINING CONFIGURATION
# =============================================================================
RANDOM_STATE = 18
TEST_SIZE    = 0.2
DATASET_PATH = 'rutile_3150.xlsx'
DATASET_NAME = 'Rutile'
FIGURES_DIR  = 'Code-2_Figures'
BEST_PARAMS  = {}

if not os.path.exists(FIGURES_DIR):
    os.makedirs(FIGURES_DIR)

# =============================================================================
# UTILITY FUNCTIONS
# =============================================================================

def save_figure(fig, figure_name):
    filepath = os.path.join(FIGURES_DIR, f"{figure_name}.svg")
    fig.savefig(filepath, format='svg', bbox_inches='tight', dpi=FIGURE_DPI)
    print(f"Figure saved: {filepath}")

def add_plot_borders(ax):
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(BORDER_LINE_WIDTH)
        spine.set_edgecolor(BORDER_COLOR)

# =============================================================================
# DATA PREPROCESSING
# =============================================================================

def preprocess_data(df, test_size=TEST_SIZE, random_state=RANDOM_STATE):
    print("Dataset columns:", df.columns.tolist())
    print("Dataset shape:", df.shape)
    if 'T(\u2103)' not in df.columns:
        print("Error: Target 'T(\u2103)' missing!")
        return None

    x = df.drop(['T(\u2103)', 'Zr'], axis=1)
    y = df['T(\u2103)']

    bins       = [y.min(), 600, 800, 1000, y.max()]
    bin_labels = ['Low-T (<600\u00b0C)', 'Mid-T (600-800\u00b0C)',
                  'High-T (800-1000\u00b0C)', 'Very High-T (>1000\u00b0C)']
    y_categories = pd.cut(y, bins=bins, labels=bin_labels, include_lowest=True)

    print("\nTemperature Category Distribution:")
    for cat, cnt in y_categories.value_counts().sort_index().items():
        print(f"  {cat}: {cnt} samples ({cnt/len(y)*100:.1f}%)")

    xtrain, xtest, ytrain, ytest = train_test_split(
        x, y, test_size=test_size, random_state=random_state, stratify=y_categories)

    print("\nTraining Set Distribution:")
    for cat in bin_labels:
        n = (pd.cut(ytrain, bins=bins, labels=bin_labels, include_lowest=True) == cat).sum()
        print(f"  {cat}: {n} ({n/len(ytrain)*100:.1f}%)")

    print("\nTest Set Distribution:")
    for cat in bin_labels:
        n = (pd.cut(ytest, bins=bins, labels=bin_labels, include_lowest=True) == cat).sum()
        print(f"  {cat}: {n} ({n/len(ytest)*100:.1f}%)")

    high_T_train = (ytrain > 1000).sum()
    high_T_test  = (ytest  > 1000).sum()
    print(f"\nHigh-T Samples (>1000\u00b0C) in Train: {high_T_train} ({high_T_train/len(ytrain)*100:.1f}%)")
    print(f"High-T Samples (>1000\u00b0C) in Test:  {high_T_test}  ({high_T_test/len(ytest)*100:.1f}%)")

    scaler        = MinMaxScaler()
    xtrain_scaled = scaler.fit_transform(xtrain)
    xtest_scaled  = scaler.transform(xtest)

    return (xtrain_scaled, xtest_scaled, ytrain, ytest,
            xtrain.columns.tolist(), xtrain.values, xtest.values, scaler)

# =============================================================================
# METRICS
# =============================================================================

def calculate_metrics(y_true, y_pred):
    r2   = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = mean_absolute_error(y_true, y_pred)
    print(f"  R2   : {r2:.4f}")
    print(f"  RMSE : {rmse:.4f}")
    print(f"  MAE  : {mae:.4f}")
    return r2, rmse, mae

def save_best_model(model, model_name, dataset_name):
    model_path = f"{model_name}_{dataset_name}.pkl"
    dump(model, model_path)
    print(f"Best model saved to: {model_path}")
    return model_path

# =============================================================================
# PLOTTING FUNCTIONS
# =============================================================================

def plot_true_vs_pred_scatter(y_train, y_train_pred, y_test, y_test_pred,
                               train_metrics, test_metrics, model_name="Model"):
    _title_fp  = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_TITLE_SIZE,
                                 weight=TITLE_FONT_WEIGHT)
    _label_fp  = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_LABEL_SIZE,
                                 weight=TRUE_PRED_SCATTER_LABEL_WEIGHT)
    _legend_fp = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_LEGEND_SIZE)
    _value_fp  = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_VALUE_SIZE)
    _tick_fp   = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_TICK_SIZE,
                                 weight='bold')

    def _draw(ax, y_true, y_pred, color, label, title, metrics):
        r2, rmse, mae = metrics
        ax.set_facecolor("#e0e0e0")
        ax.grid(color="white", linewidth=0.6)
        ax.scatter(y_true, y_pred, s=TRUE_PRED_SCATTER_POINT_SIZE, color=color,
                   alpha=SCATTER_ALPHA, edgecolor='#FFFFFF',
                   linewidth=SCATTER_EDGE_WIDTH, label=label)
        lo = min(y_true.min(), y_pred.min())
        hi = max(y_true.max(), y_pred.max())
        ax.plot([lo, hi], [lo, hi], 'k--',
                linewidth=TRUE_PRED_SCATTER_LINE_WIDTH, label='Fitted Line')
        ax.set_title(title, fontproperties=_title_fp)
        ax.set_xlabel("True T (\u00b0C)", fontproperties=_label_fp)
        ax.set_ylabel("Predicted T (\u00b0C)", fontproperties=_label_fp)
        ax.tick_params(labelsize=TRUE_PRED_SCATTER_TICK_SIZE, direction=TICK_DIRECTION,
                       length=TRUE_PRED_SCATTER_TICK_LENGTH,
                       width=TRUE_PRED_SCATTER_TICK_WIDTH, color=TICK_COLOR)
        for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_tick_fp)
        for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_tick_fp)
        leg = ax.legend(frameon=True, fancybox=False, loc='upper left', prop=_legend_fp)
        leg.get_frame().set_edgecolor('none')
        leg.get_frame().set_linewidth(0.0)
        leg.get_frame().set_facecolor('white')
        txt = f"R2 : {r2:.4f}\nRMSE: {rmse:.4f}\nMAE : {mae:.4f}"
        ax.text(0.97, 0.03, txt, transform=ax.transAxes,
                fontsize=TRUE_PRED_SCATTER_VALUE_SIZE, va='bottom', ha='right',
                bbox=dict(boxstyle='square,pad=0.20', facecolor='white',
                          edgecolor='none', alpha=0.8),
                fontproperties=_value_fp)
        add_plot_borders(ax)

    fig, ax = plt.subplots(figsize=TRUE_PRED_SCATTER_FIGSIZE)
    _draw(ax, y_train, y_train_pred, SCATTER_TRAIN_COLOR,
          'Train data', f"{model_name} - Training Set", train_metrics)
    plt.tight_layout()
    save_figure(fig, f"{model_name}_True_vs_Pred_Train")
    plt.show()

    fig, ax = plt.subplots(figsize=TRUE_PRED_SCATTER_FIGSIZE)
    _draw(ax, y_test, y_test_pred, SCATTER_TEST_COLOR,
          'Test data', f"{model_name} - Test Set", test_metrics)
    plt.tight_layout()
    save_figure(fig, f"{model_name}_True_vs_Pred_Test")
    plt.show()


def _residual_scatter_ax(ax, y_pred, residuals, color):
    _label_fp  = FontProperties(family=FONT_FAMILY, size=RESIDUAL_LABEL_SIZE,
                                 weight=RESIDUAL_LABEL_WEIGHT)
    _legend_fp = FontProperties(family=FONT_FAMILY, size=RESIDUAL_LEGEND_SIZE)
    _tick_fp   = FontProperties(family=FONT_FAMILY, size=RESIDUAL_TICK_SIZE, weight='bold')
    ax.set_facecolor("white")
    ax.grid(color="white", linewidth=0.6)
    ax.scatter(y_pred, residuals, s=RESIDUAL_POINT_SIZE, color=color,
               alpha=SCATTER_ALPHA, edgecolor='#FFFFFF', linewidth=SCATTER_EDGE_WIDTH)
    ax.axhline(0, color=ZERO_LINE_COLOR, linestyle=ZERO_LINE_STYLE,
               linewidth=ZERO_LINE_WIDTH, label='Zero Error', zorder=7)
    ax.set_xlabel("Predicted Values", fontproperties=_label_fp)
    ax.set_ylabel("Residuals",        fontproperties=_label_fp)
    ax.tick_params(labelsize=RESIDUAL_TICK_SIZE, direction=TICK_DIRECTION,
                   length=RESIDUAL_TICK_LENGTH, width=RESIDUAL_TICK_WIDTH, color=TICK_COLOR)
    for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_tick_fp)
    for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_tick_fp)
    leg = ax.legend(prop=_legend_fp, frameon=True, fancybox=False)
    leg.get_frame().set_edgecolor('none')
    leg.get_frame().set_facecolor('white')
    ax.grid(True, color="white", linewidth=0.6)
    add_plot_borders(ax)


def _residual_hist_ax(ax, residuals):
    _label_fp = FontProperties(family=FONT_FAMILY, size=RESIDUAL_LABEL_SIZE,
                                weight=RESIDUAL_LABEL_WEIGHT)
    _tick_fp  = FontProperties(family=FONT_FAMILY, size=RESIDUAL_TICK_SIZE, weight='bold')
    ax.set_facecolor("white")
    ax.hist(residuals, bins=RESIDUAL_HIST_BINS, color=RESIDUAL_HIST_COLOR,
            alpha=RESIDUAL_HIST_ALPHA, edgecolor=RESIDUAL_HIST_EDGE_COLOR,
            linewidth=RESIDUAL_HIST_EDGE_WIDTH)
    ax.set_xlabel("Residuals", fontproperties=_label_fp)
    ax.set_ylabel("Frequency", fontproperties=_label_fp)
    ax.tick_params(labelsize=RESIDUAL_TICK_SIZE, direction=TICK_DIRECTION,
                   length=RESIDUAL_TICK_LENGTH, width=RESIDUAL_TICK_WIDTH, color=TICK_COLOR)
    for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_tick_fp)
    for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_tick_fp)
    ax.locator_params(axis='x', nbins=X_NBINS)
    ax.grid(True, color="white", linewidth=0.6)
    add_plot_borders(ax)


def plot_residual_analysis(y_true, y_pred, model_name):
    residuals = y_true.values - y_pred
    color     = SCATTER_TRAIN_COLOR if "Train" in model_name else SCATTER_TEST_COLOR
    figsize   = RESIDUAL_ANALYSIS_TRAIN_FIGSIZE if "Train" in model_name \
                else RESIDUAL_ANALYSIS_TEST_FIGSIZE
    _title_fp = FontProperties(family=FONT_FAMILY, size=RESIDUAL_TITLE_SIZE,
                               weight=TITLE_FONT_WEIGHT)

    fig_scatter, ax_scatter = plt.subplots(figsize=figsize)
    ax_scatter.set_title(f"{model_name} - Residuals vs Predicted",
                         fontproperties=_title_fp)
    _residual_scatter_ax(ax_scatter, y_pred, residuals, color)
    plt.tight_layout()
    save_figure(fig_scatter, f"{model_name.replace(' ', '_')}_Residuals_vs_Predicted")
    plt.show()

    fig_hist, ax_hist = plt.subplots(figsize=figsize)
    ax_hist.set_title(f"{model_name} - Residual Distribution", fontproperties=_title_fp)
    _residual_hist_ax(ax_hist, residuals)
    plt.tight_layout()
    save_figure(fig_hist, f"{model_name.replace(' ', '_')}_Residual_Distribution")
    plt.show()

    print(f"\n{model_name} - Residual Statistics:")
    print(f"  Mean : {np.mean(residuals):.4f}")
    print(f"  Std  : {np.std(residuals):.4f}")
    print(f"  Min  : {np.min(residuals):.4f}")
    print(f"  Max  : {np.max(residuals):.4f}")


def plot_feature_importance(model, feature_names, model_name,
                             top_n=FEATURE_IMPORTANCE_TOP_N):
    importance     = model.feature_importances_
    importance_pct = (importance / importance.sum()) * 100
    indices        = np.argsort(importance_pct)[-top_n:]

    _title_fp = FontProperties(family=FONT_FAMILY, size=FEATURE_IMPORTANCE_TITLE_SIZE,
                                weight=FEATURE_IMPORTANCE_TITLE_WEIGHT)
    _label_fp = FontProperties(family=FONT_FAMILY, size=FEATURE_IMPORTANCE_LABEL_SIZE,
                                weight=FEATURE_IMPORTANCE_LABEL_WEIGHT)
    _tick_fp  = FontProperties(family=FONT_FAMILY, size=FEATURE_IMPORTANCE_TICK_SIZE,
                                weight='bold')
    _value_fp = FontProperties(family=FONT_FAMILY, size=FEATURE_IMPORTANCE_VALUE_SIZE)

    cmap   = plt.cm.get_cmap(FEATURE_IMPORTANCE_COLORMAP)
    colors = cmap(np.linspace(FEATURE_IMPORTANCE_CMAP_LOW,
                              FEATURE_IMPORTANCE_CMAP_HIGH, len(indices)))

    fig, ax = plt.subplots(figsize=FEATURE_IMPORTANCE_FIGSIZE)
    ax.set_facecolor("white")
    bars = ax.barh(range(len(indices)), importance_pct[indices], color=colors,
                   edgecolor=FEATURE_IMPORTANCE_BAR_EDGE_COLOR,
                   linewidth=FEATURE_IMPORTANCE_BAR_EDGE_WIDTH,
                   alpha=FEATURE_IMPORTANCE_BAR_ALPHA,
                   height=FEATURE_IMPORTANCE_BAR_HEIGHT)

    for i, bar in enumerate(bars):
        val = importance_pct[indices[i]]
        ax.text(val + FEATURE_IMPORTANCE_VALUE_OFFSET,
                bar.get_y() + bar.get_height() / 2,
                f'{val:.1f}', ha='left', va='center',
                fontproperties=_value_fp, color='black')

    feature_labels = [f"{feature_names[i]} --- " for i in indices]
    ax.set_yticks(range(len(indices)))
    ax.set_yticklabels(feature_labels, fontproperties=_tick_fp)
    ax.set_xlabel("Feature Importance (%)", fontproperties=_label_fp)
    ax.set_title(f"{model_name} - Feature Importance", fontproperties=_title_fp)
    ax.set_xlim(0, importance_pct[indices].max() * FEATURE_IMPORTANCE_XLIM_SCALE)
    for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_tick_fp)
    ax.grid(False)
    ax.tick_params(axis='x', which='major', direction=TICK_DIRECTION,
                   length=FEATURE_IMPORTANCE_TICK_LENGTH,
                   width=FEATURE_IMPORTANCE_TICK_WIDTH,
                   color=TICK_COLOR, bottom=True, top=False)
    ax.tick_params(axis='y', left=False)
    add_plot_borders(ax)
    fig.subplots_adjust(left=0.25, right=0.95, top=0.90, bottom=0.15)
    plt.tight_layout()
    save_figure(fig, f"{model_name}_Feature_Importance_Top{top_n}")
    plt.show()

    print(f"\n{model_name} - Top {top_n} Feature Importance:")
    print("-" * 50)
    for rank, idx in enumerate(indices[::-1], 1):
        print(f"  {rank:2d}. {feature_names[idx]:4s}: "
              f"{importance_pct[idx]:6.2f}% (raw: {importance[idx]:.6f})")
    print("-" * 50)


def plot_partial_dependence(model, X_train, feature_names, model_name="Model"):
    """
    Plot one partial dependence figure per feature in PDP_SELECTED_FEATURES.

    Styling is kept consistent with the scatter panels: same figure size,
    font properties, and tick parameters. Grid is suppressed and a black
    border is drawn explicitly because PartialDependenceDisplay inherits
    the seaborn whitegrid style by default.
    """
    print(f"\n{'='*70}")
    print(f"GENERATING PARTIAL DEPENDENCE PLOTS FOR {model_name}")
    print(f"{'='*70}\n")

    _title_fp = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_TITLE_SIZE,
                                weight=TITLE_FONT_WEIGHT)
    _label_fp = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_LABEL_SIZE,
                                weight=TRUE_PRED_SCATTER_LABEL_WEIGHT)
    _tick_fp  = FontProperties(family=FONT_FAMILY, size=TRUE_PRED_SCATTER_TICK_SIZE,
                                weight='bold')

    for feature_name in PDP_SELECTED_FEATURES:
        if feature_name not in feature_names:
            print(f"  WARNING: '{feature_name}' not found — skipping.")
            continue

        feature_index = feature_names.index(feature_name)
        fig, ax = plt.subplots(figsize=TRUE_PRED_SCATTER_FIGSIZE)

        PartialDependenceDisplay.from_estimator(
            model, X_train, features=[feature_index], ax=ax
        )

        # Style line
        for line in ax.get_lines():
            line.set_color('black')
            line.set_linewidth(TRUE_PRED_SCATTER_LINE_WIDTH)

        # No grid
        ax.grid(False)

        ax.set_title(f"{model_name} - {feature_name}", fontproperties=_title_fp)
        ax.set_xlabel(feature_name,         fontproperties=_label_fp)
        ax.set_ylabel("Partial Dependence", fontproperties=_label_fp)
        ax.tick_params(labelsize=TRUE_PRED_SCATTER_TICK_SIZE,
                       direction=TICK_DIRECTION,
                       length=TRUE_PRED_SCATTER_TICK_LENGTH,
                       width=TRUE_PRED_SCATTER_TICK_WIDTH,
                       color=TICK_COLOR)
        for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_tick_fp)
        for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_tick_fp)

        # Force black border
        for spine in ax.spines.values():
            spine.set_visible(True)
            spine.set_linewidth(BORDER_LINE_WIDTH)
            spine.set_edgecolor('black')
            spine.set_color('black')

        plt.tight_layout()
        save_figure(fig, f"{model_name}_PDP_{feature_name}")
        plt.show()


def plot_shap_enhanced(model, X_train, X_test, feature_names,
                       model_name="Model", model_type="LightGBM"):
    print(f"\n{'='*70}")
    print(f"GENERATING SHAP ANALYSIS FOR {model_name} ({model_type})")
    print(f"{'='*70}")

    try:
        if model_type in ['LightGBM', 'GBR', 'XGBoost', 'RFR']:
            if model_type == 'XGBoost':
                try:
                    explainer = shap.TreeExplainer(model,
                                                   feature_perturbation='interventional')
                except Exception:
                    explainer = shap.TreeExplainer(model)
            else:
                explainer = shap.TreeExplainer(model)
            shap_values_test  = explainer.shap_values(X_test)
            shap_values_train = explainer.shap_values(X_train[:100])
        elif model_type == 'MLP':
            print("Skipping SHAP for MLP.")
            return
        elif model_type == 'SVR':
            print("Skipping SHAP for SVR.")
            return
    except Exception as e:
        print(f"Error creating SHAP explainer for {model_name}: {e}")
        return

    shap.initjs()

    _shap_title_fp = FontProperties(family=FONT_FAMILY, size=SHAP_SUMMARY_TITLE_SIZE,
                                     weight=SHAP_SUMMARY_TITLE_WEIGHT)
    _shap_label_fp = FontProperties(family=FONT_FAMILY, size=SHAP_SUMMARY_LABEL_SIZE,
                                     weight=SHAP_SUMMARY_LABEL_WEIGHT)
    _shap_tick_fp  = FontProperties(family=FONT_FAMILY, size=SHAP_SUMMARY_TICK_SIZE,
                                     weight='bold')

    plt.figure(figsize=SHAP_SUMMARY_PLOT_FIGSIZE)
    shap.summary_plot(shap_values_test, X_test, feature_names=feature_names, show=False)
    ax, fig = plt.gca(), plt.gcf()
    ax.set_facecolor("#e0e0e0"); fig.set_facecolor("white")
    ax.grid(color="white", linewidth=1.2)
    plt.title(f"{model_name} - SHAP Feature Summary", fontproperties=_shap_title_fp, pad=15)
    ax.tick_params(labelsize=SHAP_SUMMARY_TICK_SIZE)
    for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_shap_tick_fp)
    for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_shap_tick_fp)
    ax.set_xlabel("SHAP Value (Feature Impact on Prediction)", fontproperties=_shap_label_fp)
    for cax in fig.axes:
        if cax is not ax:
            cax.tick_params(labelsize=SHAP_SUMMARY_TICK_SIZE)
            for lbl in cax.get_yticklabels(): lbl.set_fontsize(SHAP_SUMMARY_TICK_SIZE)
            cax.yaxis.label.set_fontsize(SHAP_SUMMARY_LABEL_SIZE)
            cax.yaxis.label.set_fontweight('bold')
    add_plot_borders(ax)
    plt.tight_layout()
    save_figure(fig, f"{model_name}_SHAP_Summary_Plot")
    plt.show()

    plt.figure(figsize=SHAP_BAR_PLOT_FIGSIZE)
    shap.summary_plot(shap_values_test, X_test, feature_names=feature_names,
                      plot_type="bar", show=False)
    plt.title(f"{model_name} - SHAP Feature Importance",
              fontproperties=_shap_title_fp, pad=15)
    ax = plt.gca()
    ax.tick_params(labelsize=SHAP_SUMMARY_TICK_SIZE)
    for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_shap_tick_fp)
    for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_shap_tick_fp)
    ax.set_xlabel("Mean |SHAP Value|", fontproperties=_shap_label_fp)
    add_plot_borders(ax)
    plt.tight_layout()
    save_figure(plt.gcf(), f"{model_name}_SHAP_Bar_Plot")
    plt.show()

    plt.figure(figsize=SHAP_WATERFALL_PLOT_FIGSIZE)
    sample_idx  = min(SHAP_WATERFALL_SAMPLE_IDX, len(shap_values_train) - 1)
    explanation = shap.Explanation(
        values=shap_values_train[sample_idx],
        base_values=explainer.expected_value,
        data=X_train[sample_idx],
        feature_names=feature_names
    )
    shap.waterfall_plot(explanation, show=False)
    plt.title(f"{model_name} - SHAP Waterfall Plot",
              fontproperties=FontProperties(family=FONT_FAMILY,
                                           size=SHAP_SUMMARY_TITLE_SIZE - 6,
                                           weight='bold'), pad=15)
    ax = plt.gca()
    ax.grid(False)
    _wf_tick_fp  = FontProperties(family=FONT_FAMILY,
                                   size=SHAP_SUMMARY_TICK_SIZE - 4, weight='bold')
    _wf_label_fp = FontProperties(family=FONT_FAMILY,
                                   size=SHAP_SUMMARY_LABEL_SIZE - 4, weight='bold')
    ax.tick_params(labelsize=SHAP_SUMMARY_TICK_SIZE - 4)
    for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_wf_tick_fp)
    for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_wf_tick_fp)
    ax.set_xlabel(ax.get_xlabel(), fontproperties=_wf_label_fp)
    for txt in ax.texts:
        txt.set_fontsize(SHAP_SUMMARY_TICK_SIZE - 4)
        txt.set_fontfamily(FONT_FAMILY)
    for txt in plt.gcf().texts:
        txt.set_fontsize(SHAP_SUMMARY_TICK_SIZE - 4)
        txt.set_fontfamily(FONT_FAMILY)
    add_plot_borders(ax)
    plt.tight_layout()
    save_figure(plt.gcf(), f"{model_name}_SHAP_Waterfall_Plot")
    plt.show()

    mean_abs_shap = np.abs(shap_values_test).mean(axis=0)
    sorted_idx    = np.argsort(mean_abs_shap)[::-1]
    top_feat_idxs = sorted_idx[:SHAP_TOP_N_FEATURES]

    fig, axes = plt.subplots(2, 3, figsize=SHAP_DEPENDENCE_GRID_FIGSIZE)
    axes = axes.flatten()

    for i, feat_idx in enumerate(top_feat_idxs):
        ax        = axes[i]
        feat_name = feature_names[feat_idx]
        interactions = [(j, abs(np.corrcoef(X_test[:, feat_idx],
                                             shap_values_test[:, j])[0, 1]))
                        for j in range(len(feature_names)) if j != feat_idx]
        interactions.sort(key=lambda x: x[1], reverse=True)
        inter_idx  = interactions[0][0] if interactions else feat_idx
        inter_name = feature_names[inter_idx]
        sc = ax.scatter(X_test[:, feat_idx], shap_values_test[:, feat_idx],
                        c=X_test[:, inter_idx], cmap=SHAP_COLORMAP,
                        s=SHAP_SCATTER_SIZE, alpha=SHAP_SCATTER_ALPHA,
                        edgecolors=SHAP_SCATTER_EDGE_COLOR,
                        linewidth=SHAP_SCATTER_EDGE_WIDTH)
        cbar = plt.colorbar(sc, ax=ax)
        cbar.set_label(inter_name, fontproperties=_shap_label_fp)
        cbar.ax.tick_params(labelsize=SHAP_SUMMARY_TICK_SIZE)
        for lbl in cbar.ax.get_yticklabels(): lbl.set_fontproperties(_shap_tick_fp)
        cbar.outline.set_visible(False)
        ax.set_facecolor("#f8f8f8"); ax.grid(False)
        ax.set_title(feat_name,    fontproperties=_shap_title_fp, pad=10)
        ax.set_xlabel(feat_name,   fontproperties=_shap_label_fp)
        ax.set_ylabel("SHAP Value",fontproperties=_shap_label_fp)
        ax.tick_params(labelsize=SHAP_SUMMARY_TICK_SIZE)
        for lbl in ax.get_yticklabels(): lbl.set_fontproperties(_shap_tick_fp)
        for lbl in ax.get_xticklabels(): lbl.set_fontproperties(_shap_tick_fp)
        add_plot_borders(ax)

    ax_inter = axes[5]
    top_names, top_strengths = [], []
    for idx in sorted_idx[:SHAP_TOP_N_FEATURES]:
        total = sum(abs(np.corrcoef(X_test[:, idx], shap_values_test[:, j])[0, 1])
                    for j in range(len(feature_names)) if j != idx)
        top_names.append(feature_names[idx])
        top_strengths.append(total)

    y_pos    = np.arange(len(top_names))
    i_cmap   = plt.cm.get_cmap(INTERACTION_COLORMAP)
    i_colors = i_cmap(np.linspace(INTERACTION_CMAP_LOW, INTERACTION_CMAP_HIGH,
                                   len(top_names)))
    ax_inter.barh(y_pos, top_strengths, color=i_colors,
                  alpha=INTERACTION_BAR_ALPHA,
                  edgecolor=INTERACTION_BAR_EDGE_COLOR,
                  linewidth=INTERACTION_BAR_EDGE_WIDTH,
                  height=INTERACTION_BAR_HEIGHT)
    ax_inter.set_yticks(y_pos)
    ax_inter.set_yticklabels(top_names, fontproperties=_shap_tick_fp)
    ax_inter.set_xlabel('Total Interaction Strength',
                         fontproperties=_shap_label_fp, labelpad=6)
    ax_inter.set_title('Feature Interaction Summary',
                        fontproperties=_shap_title_fp, pad=10)
    ax_inter.set_facecolor("#f8f8f8"); ax_inter.grid(False)
    ax_inter.tick_params(axis='x', labelsize=SHAP_SUMMARY_TICK_SIZE)
    ax_inter.tick_params(axis='y', labelsize=SHAP_SUMMARY_TICK_SIZE)
    for lbl in ax_inter.get_xticklabels(): lbl.set_fontproperties(_shap_tick_fp)
    ax_inter.set_xlim(0, max(top_strengths) * 1.1)
    add_plot_borders(ax_inter)

    fig.suptitle(f'{model_name} - SHAP Feature Dependence and Interactions',
                 fontproperties=_shap_title_fp, y=0.995)
    plt.tight_layout()
    save_figure(fig, f"{model_name}_SHAP_Dependence_Interactions")
    plt.show()

    print(f"\n{model_name} SHAP analysis done.")
    print("Top 5 Most Important Features (by mean |SHAP|):")
    for i, idx in enumerate(sorted_idx[:5], 1):
        print(f"  {i}. {feature_names[idx]}: {mean_abs_shap[idx]:.4f}")
    print("\nTop Feature Interactions:")
    for i, idx in enumerate(sorted_idx[:4], 1):
        interactions = [(feature_names[j],
                         abs(np.corrcoef(X_test[:, idx],
                                         shap_values_test[:, j])[0, 1]))
                        for j in range(len(feature_names)) if j != idx]
        interactions.sort(key=lambda x: x[1], reverse=True)
        strongest = interactions[0] if interactions else ("None", 0)
        print(f"  {i}. {feature_names[idx]} <- {strongest[0]}: {strongest[1]:.4f}")

# =============================================================================
# MODEL TRAINING FUNCTIONS
# =============================================================================

def train_and_save_lgbm_model(xtrain, ytrain, dataset_name):
    print("\n--- Training LightGBM Regressor ---")
    param_grid = {'num_leaves': [31], 'n_estimators': [100],
                  'learning_rate': [0.1], 'max_depth': [7]}
    lgbm = lgb.LGBMRegressor(random_state=RANDOM_STATE, n_jobs=1, objective='regression')
    grid_search = GridSearchCV(lgbm, param_grid, cv=5,
                               scoring='neg_mean_squared_error', n_jobs=1)
    grid_search.fit(xtrain, ytrain)
    best = grid_search.best_estimator_
    BEST_PARAMS["LightGBM"] = grid_search.best_params_
    save_best_model(best, 'LGBM', dataset_name)
    print(f"\nBest parameters: {grid_search.best_params_}")
    return best

def train_and_save_best_gbr_model(xtrain, ytrain, dataset_name):
    print("\n--- Training Gradient Boosting Regressor ---")
    param_grid = {'n_estimators': [200], 'learning_rate': [0.1], 'max_depth': [7],
                  'subsample': [1.0], 'max_features': [0.7], 'min_impurity_decrease': [0.0]}
    gbr = GradientBoostingRegressor(random_state=RANDOM_STATE)
    grid_search = GridSearchCV(gbr, param_grid, cv=5,
                               scoring='neg_mean_squared_error', n_jobs=1)
    grid_search.fit(xtrain, ytrain)
    best = grid_search.best_estimator_
    BEST_PARAMS["GBR"] = grid_search.best_params_
    save_best_model(best, 'GBR', dataset_name)
    print(f"\nBest parameters: {grid_search.best_params_}")
    return best

def train_and_save_xgb(xtrain, ytrain, dataset_name):
    print("\n--- Training XGBoost Regressor ---")
    param_grid = {'n_estimators': [50], 'learning_rate': [0.2], 'max_depth': [5],
                  'subsample': [0.7], 'colsample_bytree': [0.9], 'gamma': [0.50]}
    xg_reg = xgb.XGBRegressor(objective='reg:squarederror', n_jobs=1,
                               random_state=RANDOM_STATE)
    grid_search = GridSearchCV(xg_reg, param_grid, cv=5,
                               scoring='neg_mean_squared_error', n_jobs=1)
    grid_search.fit(xtrain, ytrain)
    best = grid_search.best_estimator_
    BEST_PARAMS["XGBoost"] = grid_search.best_params_
    save_best_model(best, 'XGBoost', dataset_name)
    print(f"\nBest parameters: {grid_search.best_params_}")
    return best

def train_and_save_best_rfr_model(xtrain, ytrain, dataset_name):
    print("\n--- Training Random Forest Regressor ---")
    param_grid = {'n_estimators': [200], 'max_depth': [10], 'min_samples_split': [2],
                  'min_samples_leaf': [2], 'max_features': [0.8]}
    rfr = RandomForestRegressor(random_state=RANDOM_STATE)
    grid_search = GridSearchCV(rfr, param_grid, cv=5,
                               scoring='neg_mean_squared_error', n_jobs=1)
    grid_search.fit(xtrain, ytrain)
    best = grid_search.best_estimator_
    BEST_PARAMS["RFR"] = grid_search.best_params_
    save_best_model(best, 'RFR', dataset_name)
    print(f"\nBest parameters: {grid_search.best_params_}")
    return best

def train_and_save_best_mlp_model(xtrain, ytrain, dataset_name):
    print("\n--- Training MLP Regressor ---")
    param_grid = {'hidden_layer_sizes': (100, 50), 'activation': ['tanh'],
                  'alpha': [0.01], 'solver': ['adam']}
    mlp = MLPRegressor(random_state=RANDOM_STATE, max_iter=2000)
    grid_search = GridSearchCV(estimator=mlp, param_grid=param_grid, cv=5,
                               scoring='neg_mean_squared_error', n_jobs=1, verbose=0)
    grid_search.fit(xtrain, ytrain)
    best = grid_search.best_estimator_
    BEST_PARAMS["MLP"] = grid_search.best_params_
    save_best_model(best, 'MLP', dataset_name)
    print(f"\nBest parameters: {grid_search.best_params_}")
    return best

def train_and_save_best_svr_model(xtrain, ytrain, dataset_name):
    print("\n--- Training Support Vector Regressor (SVR) ---")
    param_grid = [{'kernel': ['rbf'], 'C': [1000], 'gamma': ['scale'], 'epsilon': [0.2]}]
    svr = SVR(max_iter=10000)
    grid_search = GridSearchCV(svr, param_grid, cv=5,
                               scoring='neg_mean_squared_error', n_jobs=1, verbose=1)
    grid_search.fit(xtrain, ytrain)
    best = grid_search.best_estimator_
    BEST_PARAMS["SVR"] = grid_search.best_params_
    save_best_model(best, 'SVR', dataset_name)
    print(f"\nBest parameters: {grid_search.best_params_}")
    print(f"Best CV RMSE: {np.sqrt(-grid_search.best_score_):.4f}")
    return best

# =============================================================================
# MAIN
# =============================================================================

def main():
    try:
        print(f"Loading dataset: {DATASET_PATH}")
        df = pd.read_excel(DATASET_PATH)
        print(f"Loaded dataset shape: {df.shape}")
    except FileNotFoundError:
        print(f"ERROR: Dataset file not found at {DATASET_PATH}.")
        return

    preprocessed = preprocess_data(df, test_size=TEST_SIZE, random_state=RANDOM_STATE)
    if preprocessed is None:
        return

    (x_train_scaled, x_test_scaled, y_train, y_test,
     feature_names, x_train_data_array, x_test_data_array, scaler) = preprocessed

    scaler_path = f'Scaler_{DATASET_NAME}.pkl'
    dump(scaler, scaler_path)
    print(f"\n{'='*70}")
    print(f"SCALER SAVED: {scaler_path}")
    print(f"  Features: {feature_names}")
    print(f"{'='*70}\n")

    # ── LightGBM ──────────────────────────────────────────────────────────────
    lgbm_best = train_and_save_lgbm_model(x_train_scaled, y_train, DATASET_NAME)
    print("\n" + "--- LGBM TRAINING SET EVALUATION ---".center(70, '-'))
    y_train_pred_lgbm  = lgbm_best.predict(x_train_scaled)
    train_metrics_lgbm = calculate_metrics(y_train, y_train_pred_lgbm)
    print("\n" + "--- LGBM TEST SET EVALUATION ---".center(70, '-'))
    y_test_pred_lgbm   = lgbm_best.predict(x_test_scaled)
    test_metrics_lgbm  = calculate_metrics(y_test, y_test_pred_lgbm)
    plot_true_vs_pred_scatter(y_train, y_train_pred_lgbm, y_test, y_test_pred_lgbm,
                               train_metrics_lgbm, test_metrics_lgbm, model_name="LightGBM")
    plot_residual_analysis(y_train, y_train_pred_lgbm, "LightGBM (Train)")
    plot_residual_analysis(y_test,  y_test_pred_lgbm,  "LightGBM (Test)")
    plot_feature_importance(lgbm_best, feature_names, "LightGBM")
    plot_shap_enhanced(lgbm_best, x_train_scaled, x_test_scaled,
                       feature_names, model_name="LightGBM", model_type="LightGBM")
    plot_partial_dependence(lgbm_best, x_train_scaled, feature_names, model_name="LightGBM")

    # ── GBR ───────────────────────────────────────────────────────────────────
    gbr_best = train_and_save_best_gbr_model(x_train_scaled, y_train, DATASET_NAME)
    print("\n" + "--- GBR TRAINING SET EVALUATION ---".center(70, '-'))
    y_train_pred_gbr  = gbr_best.predict(x_train_scaled)
    train_metrics_gbr = calculate_metrics(y_train, y_train_pred_gbr)
    print("\n" + "--- GBR TEST SET EVALUATION ---".center(70, '-'))
    y_test_pred_gbr   = gbr_best.predict(x_test_scaled)
    test_metrics_gbr  = calculate_metrics(y_test, y_test_pred_gbr)
    plot_true_vs_pred_scatter(y_train, y_train_pred_gbr, y_test, y_test_pred_gbr,
                               train_metrics_gbr, test_metrics_gbr, model_name="GBR")
    plot_residual_analysis(y_train, y_train_pred_gbr, "GBR (Train)")
    plot_residual_analysis(y_test,  y_test_pred_gbr,  "GBR (Test)")
    plot_feature_importance(gbr_best, feature_names, "GBR")
    plot_shap_enhanced(gbr_best, x_train_scaled, x_test_scaled,
                       feature_names, model_name="GBR", model_type="GBR")
    plot_partial_dependence(gbr_best, x_train_scaled, feature_names, model_name="GBR")

    # ── XGBoost ───────────────────────────────────────────────────────────────
    xgb_best = train_and_save_xgb(x_train_scaled, y_train, DATASET_NAME)
    print("\n" + "--- XGBoost TRAINING SET EVALUATION ---".center(70, '-'))
    y_train_pred_xgb  = xgb_best.predict(x_train_scaled)
    train_metrics_xgb = calculate_metrics(y_train, y_train_pred_xgb)
    print("\n" + "--- XGBoost TEST SET EVALUATION ---".center(70, '-'))
    y_test_pred_xgb   = xgb_best.predict(x_test_scaled)
    test_metrics_xgb  = calculate_metrics(y_test, y_test_pred_xgb)
    plot_true_vs_pred_scatter(y_train, y_train_pred_xgb, y_test, y_test_pred_xgb,
                               train_metrics_xgb, test_metrics_xgb, model_name="XGBoost")
    plot_residual_analysis(y_train, y_train_pred_xgb, "XGBoost (Train)")
    plot_residual_analysis(y_test,  y_test_pred_xgb,  "XGBoost (Test)")
    plot_feature_importance(xgb_best, feature_names, "XGBoost")
    plot_shap_enhanced(xgb_best, x_train_scaled, x_test_scaled,
                       feature_names, model_name="XGBoost", model_type="XGBoost")
    plot_partial_dependence(xgb_best, x_train_scaled, feature_names, model_name="XGBoost")

    # ── Random Forest ─────────────────────────────────────────────────────────
    rfr_best = train_and_save_best_rfr_model(x_train_scaled, y_train, DATASET_NAME)
    print("\n" + "--- RFR TRAINING SET EVALUATION ---".center(70, '-'))
    y_train_pred_rfr  = rfr_best.predict(x_train_scaled)
    train_metrics_rfr = calculate_metrics(y_train, y_train_pred_rfr)
    print("\n" + "--- RFR TEST SET EVALUATION ---".center(70, '-'))
    y_test_pred_rfr   = rfr_best.predict(x_test_scaled)
    test_metrics_rfr  = calculate_metrics(y_test, y_test_pred_rfr)
    plot_true_vs_pred_scatter(y_train, y_train_pred_rfr, y_test, y_test_pred_rfr,
                               train_metrics_rfr, test_metrics_rfr, model_name="RFR")
    plot_residual_analysis(y_train, y_train_pred_rfr, "RFR (Train)")
    plot_residual_analysis(y_test,  y_test_pred_rfr,  "RFR (Test)")
    plot_feature_importance(rfr_best, feature_names, "RFR")
    plot_shap_enhanced(rfr_best, x_train_scaled, x_test_scaled,
                       feature_names, model_name="RFR", model_type="RFR")
    plot_partial_dependence(rfr_best, x_train_scaled, feature_names, model_name="RFR")

    # ── MLP ───────────────────────────────────────────────────────────────────
    mlp_best = train_and_save_best_mlp_model(x_train_scaled, y_train, DATASET_NAME)
    print("\n" + "--- MLP TRAINING SET EVALUATION ---".center(70, '-'))
    y_train_pred_mlp  = mlp_best.predict(x_train_scaled)
    train_metrics_mlp = calculate_metrics(y_train, y_train_pred_mlp)
    print("\n" + "--- MLP TEST SET EVALUATION ---".center(70, '-'))
    y_test_pred_mlp   = mlp_best.predict(x_test_scaled)
    test_metrics_mlp  = calculate_metrics(y_test, y_test_pred_mlp)
    plot_true_vs_pred_scatter(y_train, y_train_pred_mlp, y_test, y_test_pred_mlp,
                               train_metrics_mlp, test_metrics_mlp, model_name="MLP")
    plot_residual_analysis(y_train, y_train_pred_mlp, "MLP (Train)")
    plot_residual_analysis(y_test,  y_test_pred_mlp,  "MLP (Test)")

    # ── SVR ───────────────────────────────────────────────────────────────────
    svr_best = train_and_save_best_svr_model(x_train_scaled, y_train, DATASET_NAME)
    print("\n" + "--- SVR TRAINING SET EVALUATION ---".center(70, '-'))
    y_train_pred_svr  = svr_best.predict(x_train_scaled)
    train_metrics_svr = calculate_metrics(y_train, y_train_pred_svr)
    print("\n" + "--- SVR TEST SET EVALUATION ---".center(70, '-'))
    y_test_pred_svr   = svr_best.predict(x_test_scaled)
    test_metrics_svr  = calculate_metrics(y_test, y_test_pred_svr)
    plot_true_vs_pred_scatter(y_train, y_train_pred_svr, y_test, y_test_pred_svr,
                               train_metrics_svr, test_metrics_svr, model_name="SVR")
    plot_residual_analysis(y_train, y_train_pred_svr, "SVR (Train)")
    plot_residual_analysis(y_test,  y_test_pred_svr,  "SVR (Test)")

    print("\n" + "="*70)
    print("All 6 models trained and evaluated.")
    print("="*70)
    print("\nBest hyperparameters for each model:")
    for mn, params in BEST_PARAMS.items():
        print(f"\n{mn}:")
        for k, v in params.items():
            print(f"  {k}: {v}")


if __name__ == "__main__":
    main()
```
