```python
import pandas as pd
from joblib import load

# Load dataset, trained model, and fitted scaler
new_data = pd.read_excel('Rutile-DS-3.xlsx')
model = load('RFR_Rutile_DS2.pkl')
scaler = load('Scaler_Rutile_DS2.pkl')  

# Subset to the four features the model was trained on
feature_cols = ['Zr', 'Hf', 'Th', 'U']
X_new = new_data[feature_cols]

# Scale features using the scaler fitted on the training set
X_scaled = scaler.transform(X_new)

# Run inference
predictions = model.predict(X_scaled)

# Append predictions and write output
new_data['Predicted_T'] = predictions
new_data.to_excel('Prediction_results.xlsx', index=False)

print(f"Predicted {len(predictions)} samples.")
print(f"Results saved to: Prediction_results.xlsx")
```
